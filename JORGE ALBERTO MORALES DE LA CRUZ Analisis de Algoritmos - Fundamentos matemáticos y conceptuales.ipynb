- *¿Qué establece el Principio de Invarianza respecto al tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo?*  
**RESPUESTA:**  
El Principio de Invarianza establece los siguientes puntos clave respecto al tiempo de ejecución de dos implementaciones distintas de un mismo algoritmo:

* **Diferencia acotada:** El tiempo de ejecución entre dos implementaciones (ya sea en máquinas distintas o con códigos diferentes) no diferirá por más de una constante multiplicativa.


* **Independencia de la implementación:** El tiempo que requiere un algoritmo para resolver un problema depende de la naturaleza intrínseca del algoritmo y no de los detalles particulares de su implementación.


* **Definición matemática:** Si se tienen dos implementaciones que tardan $T_1(n)$ y $T_2(n)$ respectivamente, el principio asegura que existe una constante real $c > 0$ y un número natural $n_0$ tales que para todo $n \ge n_0$ se verifica que $T_1(n) \le c \cdot T_2(n)$.

* La función de techo (ceiling) de un número real *x* obtiene el menor entero que sea mayor o igual a *x*.

Según el texto, se establece que una función exponencial positiva crece más rápido que un polinomio debido a la siguiente demostración matemática:

* Para todo valor de *n* y una constante *a* ≥ 1, se cumple que el límite de $n^b / a^n$ cuando *n* tiende a infinito es igual a 0 ($\lim_{n \to \infty} n^b/a^n = 0$).

**RESPUESTA:**  
Basado en los documentos proporcionados, aquí tienes las respuestas a tus preguntas:

**Definición y características de un algoritmo**
Un algoritmo es un conjunto ordenado y finito de instrucciones precisas que permiten resolver un problema mediante la transformación de una entrada en una salida. Sus características clave son:

* **Entrada y Salida:** Debe tener claramente definidos los datos o valores que necesita para operar (entrada) y los que debe producir (salida).


* **Preciso y no ambiguo:** Debe indicar el orden exacto de ejecución de manera explícita y sin ambigüedades, de tal forma que siempre se sepa qué paso sigue.


* **Finito:** Debe finalizar después de una secuencia o número finito de pasos.


* **Correcto:** Debe hacer lo que se supone que debe hacer, generando la solución correcta al problema.


* **Efectividad:** Cada instrucción debe poder completarse en un tiempo finito y ser lo suficientemente básica para ser ejecutada.


* **General:** Debe ser lo suficientemente general para poder contemplar todos los casos posibles de entrada.



**El Problema de Paro (Halting Problem)**
Formulado por Alan Turing, el "Problema de Paro" plantea la siguiente interrogante: dado un programa de computadora y una entrada específica, ¿es posible determinar si el programa se detendrá en algún momento o si se quedará en un ciclo infinito?. Turing demostró que este es un problema matemáticamente insoluble, es decir, comprobó que no existe ningún algoritmo o programa de computadora capaz de analizar a todos los demás programas y predecir correctamente si se detendrán para todas sus posibles instancias de entrada.

**Análisis A Priori vs. Prueba A Posteriori**
Ambos son métodos para evaluar la eficiencia (tiempo de ejecución y uso de memoria) de los algoritmos, pero se realizan en etapas y con enfoques distintos:

* **Análisis A Priori (teórico):** Se lleva a cabo en la etapa de diseño, antes de implementar el código. Su objetivo es obtener una función o expresión matemática que describa el comportamiento del algoritmo y limite su tiempo de cálculo en función de los parámetros de entrada. Este análisis es útil para predecir el costo del algoritmo y evitar implementaciones innecesariamente laboriosas.


* **Prueba A Posteriori (empírica o experimental):** Se realiza una vez que el algoritmo ha sido programado. Consiste en ejecutar el programa en una computadora utilizando casos de prueba reales para medir y recoger estadísticas exactas sobre el tiempo y el espacio de memoria que realmente consumió durante su ejecución.

- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**  
* **Peor caso:** Representa la situación en la cual el algoritmo consume la mayor cantidad de tiempo o ejecuta el mayor número de instrucciones para resolver un problema. Este caso indica el tiempo máximo obtenido tras considerar todas las entradas posibles para un tamaño de problema determinado. Su análisis es fundamental porque proporciona una cota superior del trabajo efectuado, garantizando que bajo ninguna circunstancia el algoritmo tardará más que ese límite. Por esta seguridad que brinda, es el escenario que se utiliza casi siempre para medir y comparar la eficiencia general de los algoritmos.


* **Mejor caso:** Corresponde a la situación opuesta, en la que el algoritmo consume la menor cantidad de tiempo o el menor número de instrucciones para llegar a la solución. Señala el tiempo mínimo de ejecución al evaluar todas las entradas posibles. Un ejemplo clásico se da en los algoritmos de ordenamiento cuando los datos de entrada ya se encuentran completamente ordenados desde el inicio. Aunque su análisis suele ser el más fácil de realizar, muchas veces se considera una medida poco realista o engañosa para predecir el rendimiento general.


* **Caso promedio:** Indica el tiempo medio obtenido o el comportamiento típico esperado cuando el algoritmo se ejecuta muchas veces sobre una gran variedad de datos. Es considerado la medida más realista del desempeño de un algoritmo, pero también es el más difícil de calcular matemáticamente. Para poder determinar este valor, es estrictamente necesario conocer o asumir cómo están distribuidos los datos de entrada y asignar una probabilidad a cada escenario para calcular un promedio ponderado.

### Instrucciones:
1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
# factorial recursivo
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)


# fibonacci iterativo
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        a, b = b, a + b

    return a

n = 5

print("Factorial de", n, ":", factorial(n))
print("Fibonacci de", n, ":", fibonacci(n))
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
# fibonacci recursivo
def fibonacci_recursivo(n):
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

# fibonacci iterativo
def fibonacci_iterativo(n):
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b
    return a

n = 35

print("Fibonacci recursivo:")
%timeit fibonacci_recursivo(n)

print("\nFibonacci iterativo:")
%timeit fibonacci_iterativo(n)
````
